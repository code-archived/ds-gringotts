```python
import pandas as pd
import tablescraper as ts
```

## Wikipedia Tables

```python
tables = ts.wikitable(weburl = "https://en.wikipedia.org/wiki/List_of_cities_in_India_by_population", verbose = True)
tables[0].sample(3)
```

## Chittorgarh Website

```python
table = ts.chittorgarh(
    weburl = "https://www.chittorgarh.com/report/ipo-in-india-list-main-board-sme/82/mainboard/",
    parsedates = True, datecolumns = ["OpeningDate", "ClosingDate", "ListingDate"],
    columns = ["CompanyName", "OpeningDate", "ClosingDate", "ListingDate", "ListingExchange"], dropindex = 0,
    dropcols = ["Issue Price (Rs.)▲▼", "Total Issue Amount (Incl.Firm reservations) (Rs.cr.)▲▼", "Lead Manager▲▼", "Compare"]
)
table.sample(3)
```
